```python
from nilearn import plotting, image
from pathlib import Path
import re
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# --- Path to results directory ---
results_dir = Path("/Users/silvycollin/Documents/research/narrativeProject/processed_first_level_per_sentence/tunnel/7_MAC/SecondLevel_contrast/VsBaseline/")

# --- Load all .nii or .nii.gz files ---
contrast_files = list(results_dir.glob("*.nii*"))

# --- Extract number after 'foundation' and sort files ---
def get_foundation_number(filename):
    match = re.search(r'foundation(\d+)', filename.lower())
    return int(match.group(1)) if match else float('inf')

contrast_files_sorted = sorted(contrast_files, key=lambda f: get_foundation_number(f.stem))

# --- Load images ---
imgs = [image.load_img(f) for f in contrast_files_sorted]

# --- Titles from two words after '_SecondLevel_' ---
titles = []
for f in contrast_files_sorted:
    stem = f.stem
    try:
        parts = stem.split("_SecondLevel_")[1].split("_")
        title_text = " ".join([p.capitalize() for p in parts[:2]])
    except IndexError:
        title_text = stem
    titles.append(title_text)

# --- Determine global vmax for consistent color scale ---
vmax = max(np.max(img.get_fdata()[img.get_fdata() > 0]) for img in imgs)
vmin = 0

# --- Layout parameters ---
n_orientations = 3  # sagittal, coronal, axial
fig_height_per_row = 2.5
fig = plt.figure(figsize=(6*n_orientations, fig_height_per_row*len(imgs)))
gs = GridSpec(len(imgs), n_orientations, figure=fig, hspace=0.3, wspace=0.1)

# --- Slice positions per orientation ---
def get_slice_coords(img, n_slices=1):
    data = img.get_fdata()
    coords = []
    for axis in range(3):
        non_zero = np.where(data > 0)
        if len(non_zero[axis]) == 0:
            slice_idx = [data.shape[axis] // 2]
        else:
            slice_idx = [int(np.median(non_zero[axis]))]
        coords.append(slice_idx)
    return coords  # [sagital, coronal, axial]

# --- Plot each contrast ---
for i, img in enumerate(imgs):
    slice_coords = get_slice_coords(img)
    for j, axis in enumerate(['x', 'y', 'z']):  # sagittal, coronal, axial
        ax = fig.add_subplot(gs[i, j])
        plotting.plot_stat_map(
            img,
            display_mode=axis,
            cut_coords=slice_coords[j],
            threshold=0,
            vmax=vmax,
            vmin=vmin,
            cmap='hot',
            axes=ax,
            colorbar=False,
            annotate=False
        )
        if j == 1:  # title above middle slice
            ax.set_title(titles[i], fontsize=10, pad=3)

# --- Add horizontal colorbar ---
cbar_ax = fig.add_axes([0.15, 0.02, 0.7, 0.03])
sm = plt.cm.ScalarMappable(cmap='hot', norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm.set_array([])
fig.colorbar(sm, cax=cbar_ax, orientation='horizontal', label='Activation (positive only)')

plt.show()
```

```python
from nilearn import image, plotting
from pathlib import Path
import numpy as np
import re
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# --- Path to results directory ---
results_dir = Path("/Users/silvycollin/Documents/research/narrativeProject/processed_first_level_per_sentence/tunnel/7_MAC/SecondLevel_contrast/VsBaseline/")

# --- Load and sort images by 'foundation' number ---
contrast_files = list(results_dir.glob("*.nii*"))

def get_foundation_number(filename):
    match = re.search(r'foundation(\d+)', filename.lower())
    return int(match.group(1)) if match else float('inf')

contrast_files_sorted = sorted(contrast_files, key=lambda f: get_foundation_number(f.stem))
imgs = [image.load_img(f) for f in contrast_files_sorted]

# --- Binarize images (positive values -> 1, else 0) ---
bin_imgs = []
for img in imgs:
    data = img.get_fdata()
    binary_data = (data > 0).astype(np.int32)
    bin_img = image.new_img_like(img, binary_data)
    bin_imgs.append(bin_img)

# --- Sum binary images (counts how many have activation) ---
sum_img = image.math_img(
    " + ".join([f"img{i+1}" for i in range(len(bin_imgs))]),
    **{f"img{i+1}": bin_imgs[i] for i in range(len(bin_imgs))}
)

# --- Mask zeros (make transparent) ---
sum_data = sum_img.get_fdata()
sum_data[sum_data == 0] = np.nan
sum_img_masked = image.new_img_like(sum_img, sum_data)

# --- Discrete warm color map for 1–7 ---
#colors = plt.cm.hot(np.linspace(0.3, 1, 7))
colors = plt.cm.YlOrRd(np.linspace(0.3, 1, 7))

cmap = ListedColormap(colors)

# --- Plot overlap image ---
fig = plt.figure(figsize=(8, 6))
display = plotting.plot_glass_brain(
    sum_img_masked,
    threshold=0.01,
    colorbar=False,
    display_mode='lyrz',  # left, right, sagittal, axial
    plot_abs=False,
    cmap=cmap,
    title="Overlap across 7 contrasts",
    figure=fig
)

# --- Add discrete horizontal colorbar ---
cbar_ax = fig.add_axes([0.25, 0.05, 0.5, 0.03])  # [left, bottom, width, height]
sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=1, vmax=7))
sm.set_array(np.arange(1, 8))
cbar = fig.colorbar(sm, cax=cbar_ax, orientation='horizontal', ticks=np.arange(1, 8))
cbar.set_label('Number of contrasts', fontsize=10)

plt.tight_layout()
plt.show()
```
