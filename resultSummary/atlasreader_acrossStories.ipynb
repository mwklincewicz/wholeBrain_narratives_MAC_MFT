#### code that binarizes each .nii.gz contrast result file (fdr corrected ones) to 1 for fdr corrected voxels and 0 for the rest, then for each foundation separately summarizes how many stories have that voxels fdr corrected active

```python

output_dir = Path("/Users/silvycollin/Documents/research/narrativeProject/processed_first_level_per_sentence/output_summarizedAcrossStories/")

name_title = "virtues"
# name_title = "vices"



```

```python
for foundation in [1,2,3,4,5,6,7]:

    if name_title == "virtues":
        results_dir = Path("/Users/silvycollin/Documents/research/narrativeProject/processed_first_level_per_sentence/virtues/")
        output_path = f"sum_overlap_all_contrasts_virtues_foundation{foundation}.nii.gz"
        print('run virtues')

    if name_title == "vices":
        results_dir = Path("/Users/silvycollin/Documents/research/narrativeProject/processed_first_level_per_sentence/vices/")
        output_path = f"sum_overlap_all_contrasts_vices_foundation{foundation}.nii.gz"
        print('run vices')
    
    # Load only files with the chosen foundation number 
    contrast_files = sorted(
        f for f in results_dir.glob("*.nii*")
        if re.search(fr'foundation{foundation}', f.stem.lower())
    )
    
    print(f"Found {len(contrast_files)} contrast maps with foundation{foundation}.")
    
    # Fetch MNI152 template as reference 
    mni152 = datasets.load_mni152_template()
    ref_img = mni152
    print("Reference MNI152 shape:", ref_img.shape)
    
    # Initialize sum array 
    sum_data = np.zeros(ref_img.shape, dtype=np.int32)
    
    #  Iterate through all  
    for f in contrast_files:
        img = image.load_img(f)
        # Resample
        img_resampled = image.resample_to_img(img, ref_img, interpolation='nearest',force_resample=True,copy_header=True)
        data = img_resampled.get_fdata()
        
        # Binarize: positive 1, else 0
        binary_data = (data > 0).astype(np.int32)
        sum_data += binary_data
    
    sum_img = image.new_img_like(ref_img, sum_data)
    
    sum_img.to_filename(output_path)
    
    print(" Saved summed overlap image to:", output_path)
    
    # transparency
    sum_data = sum_img.get_fdata()
    sum_data[sum_data == 0] = np.nan
    sum_img_masked = image.new_img_like(sum_img, sum_data)
    
    create_output(output_path,cluster_extent=1,direction="pos",outdir=str(output_dir))
```
